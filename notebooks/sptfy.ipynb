## Tanımsal İstatistik 4: Sanatçı Frekansları

Sanatçı bazında yapılan incelemede veri setinde en fazla temsil edilen sanatçının 85 şarkı ile İbrahim Tatlıses olduğu görülmektedir. İbrahim Tatlıses'i sırasıyla Sezen Aksu (72), Ebru Gündeş (70), Tarkan (68) ve Ajda Pekkan (59) takip etmektedir.

Sonuçlar, veri setinde bazı sanatçıların diğerlerine kıyasla daha yoğun temsil edildiğini göstermektedir. Özellikle ilk beş sanatçının toplam şarkı sayısı içerisindeki payı oldukça yüksektir ve veri setinin müzikal yapısında belirgin bir ağırlığa sahiptir.

```python
Q1 = df["ŞARKI SÜRESİ(SANİYE)"].quantile(0.25)
Q3 = df["ŞARKI SÜRESİ(SANİYE)"].quantile(0.75)

IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

aykiri = df[
    (df["ŞARKI SÜRESİ(SANİYE)"] < alt_sinir)
    | (df["ŞARKI SÜRESİ(SANİYE)"] > ust_sinir)
]

print("Aykırı gözlem sayısı:", len(aykiri))
```

## Tanımsal İstatistik 5: Aykırı Değer Analizi
